```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_classification
from scipy.cluster.hierarchy import dendrogram , linkage, fcluster
```

```python
X, _ = make_classification(
    n_samples = 300,
    n_classes = 2,
    random_state = 34
)
```

```python
hierarchy1 = linkage(X, "single")
```

```python
plt.figure(figsize=(8,6))
dendrogram(
    hierarchy1,
    orientation ="top",
    show_leaf_counts =True,
    distance_sort ="descending"
)
plt.title("Make Classification")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```

```python
hierarchy2 = linkage(X, "complete")
clusters = fcluster(
    hierarchy2,
    10,
    criterion = "distance"
)
print("Number of Clusters Created: ", len(np.unique(clusters)))
```

```python
plt.figure(figsize=(8,6))
dendrogram(
    hierarchy2,
    orientation ="top",
    distance_sort ="descending",
    show_leaf_counts =True

)
plt.title("Agglomerative Clustering")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```
